```python
# Install required libraries
!pip install statsmodels

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import statsmodels.api as sm
from google.colab import files

# Upload file
uploaded = files.upload()

# Load dataset
df = pd.read_csv(list(uploaded.keys())[0])
df.head()
```

```python
print(df.info())
print("Shape:", df.shape)

# Essential columns
essential = ['mpg', 'weight', 'horsepower', 'displacement', 'acceleration', 'cylinders']
print("Essential Columns:", essential)
```

```python
# Replace '?' with NaN, convert horsepower to numeric
df['horsepower'].replace('?', np.nan, inplace=True)
df['horsepower'] = pd.to_numeric(df['horsepower'])

print("Missing values per column:\n", df.isnull().sum())
df.dropna(inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

print("Shape after cleaning:", df.shape)
df.head()
```

```python
desc_stats = df[['mpg','weight','horsepower','acceleration','displacement']].describe()
range_vals = desc_stats.loc['max'] - desc_stats.loc['min']
print("Descriptive Stats:\n", desc_stats)
print("\nRange:\n", range_vals)
```

```python
plt.hist(df['mpg'])
plt.title("Histogram of MPG")
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.show()

sns.boxplot(x='cylinders', y='mpg', data=df)
plt.title("MPG vs Cylinders")
plt.show()

plt.hist(df['horsepower'])
plt.title("Histogram of Horsepower")
plt.show()
```

```python
def remove_outliers(col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return df[(df[col] >= lower) & (df[col] <= upper)]

for col in ['weight', 'horsepower', 'acceleration']:
    plt.boxplot(df[col])
    plt.title(f"Before Outlier Removal: {col}")
    plt.show()

df = remove_outliers('weight')
df = remove_outliers('horsepower')
df = remove_outliers('acceleration')

for col in ['weight', 'horsepower', 'acceleration']:
    plt.boxplot(df[col])
    plt.title(f"After Outlier Removal: {col}")
    plt.show()
```

```python
sm.qqplot(df['mpg'], line='45')
plt.title("Q-Q Plot for MPG")
plt.show()
```

```python
corr = df[['mpg','weight','horsepower','acceleration','displacement']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

print("Correlations with MPG (descending):\n", corr['mpg'].sort_values(ascending=False))
```

```python
mpg_mean = df['mpg'].mean()
mpg_se = stats.sem(df['mpg'])
ci = stats.t.interval(0.95, len(df['mpg'])-1, mpg_mean, mpg_se)
print("95% CI for MPG:", ci)
```

```python
hp_mean = df['horsepower'].mean()
hp_se = stats.sem(df['horsepower'])

for conf in [0.90, 0.95, 0.99]:
    t_val = stats.t.ppf((1+conf)/2, len(df)-1)
    moe = t_val * hp_se
    print(f"Margin of Error ({int(conf*100)}%): {moe}")
```

```python
us = df[df['origin']==1]['mpg']
japan = df[df['origin']==3]['mpg']

t,p = stats.ttest_ind(us, japan, equal_var=False)
print("t-statistic:", t, "p-value:", p)
```

```python
print("Pearson correlation:", stats.pearsonr(df['weight'], df['mpg']))

sns.regplot(x='weight', y='mpg', data=df)
plt.title("Weight vs MPG")
plt.show()
```

```python
X = df[['weight']]
y = df['mpg']

model = LinearRegression()
model.fit(X,y)

print("Intercept β0:", model.intercept_)
print("Slope β1:", model.coef_[0])

y_pred = model.predict(X)
print("R² Score:", r2_score(y, y_pred))
```
